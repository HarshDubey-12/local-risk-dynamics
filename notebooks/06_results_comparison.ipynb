# Results Comparison

Summarize and visualize results obtained from running experiments.  This
notebook can ingest output CSVs created by `runner.save_results` and produce
comparative plots.

```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

# assume a results CSV exists from a previous run
df = pd.read_csv('results.csv') if Path('results.csv').exists() else pd.DataFrame()

if not df.empty:
    display(df.head())
    df.plot(x='model', y=['rmse','mae'], kind='bar')
    plt.title('Metrics comparison');
    plt.show()
else:
    print('Run an experiment and save results to results.csv to view them here.')
```
