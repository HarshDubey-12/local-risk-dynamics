# SGD Monte Carlo

Demonstrate training a linear model using stochastic gradient descent with
Monte Carlo subsampling.  Compare convergence and variance of estimates.

```python
import numpy as np
import matplotlib.pyplot as plt
from src.models.mc_linear import MonteCarloLinearRegression
from src.utils.sampling import subsample_indices
from src.data.loader import load_fama_french
from src.data.preprocessing import time_series_split

# load dataset and split
X,y,dates = load_fama_french('data/raw/F-F_Research_Data_5_Factors_2x3.csv')
X_train,X_test,y_train,y_test,_,_ = time_series_split(X,y,dates,0.8)

# show sampling helper usage
indices = subsample_indices(X_train.shape[0], 50, random_state=0)
print('example subsample indices', indices[:5])

mc = MonteCarloLinearRegression(n_simulations=100, subsample_size=50, random_state=0)
mc.fit(X_train, y_train)
print('Prediction sample', mc.predict(X_test[:5]))

# coefficient distribution
coefs = mc.coefs_
plt.hist(coefs[:,0], bins=20)
plt.title('Distribution of first coefficient across simulations')
plt.show()
```
