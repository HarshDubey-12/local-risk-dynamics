# LWLR Deterministic Example

A notebook showcasing a deterministic run of the locally weighted linear
regression (no Monte Carlo).  Compare predictions to the global model.

```python
import sys
from pathlib import Path

cwd = Path.cwd()
root = cwd if (cwd / 'src').exists() else cwd.parent
sys.path.insert(0, str(root))
```

```python
import numpy as np
import matplotlib.pyplot as plt
from src.models.lwlr import LocallyWeightedLinearRegression

# generate synthetic data
np.random.seed(1)
X = np.linspace(0,10,100).reshape(-1,1)
y = np.sin(X).ravel() + np.random.normal(scale=0.2, size=100)

# fit lwlr and global for comparison
lwlr = LocallyWeightedLinearRegression(bandwidth=0.5).fit(X,y)

X_plot = np.linspace(0,10,200).reshape(-1,1)
y_pred = lwlr.predict(X_plot)

plt.scatter(X, y, alpha=0.3, label='data')
plt.plot(X_plot, y_pred, color='red', label='LWLR fit')
plt.legend()
plt.title('Deterministic LWLR example')
plt.show()
```
