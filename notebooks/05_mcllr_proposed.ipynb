# Proposed MCLLR Model

Notebook for the proposed Monte Carlo Local Linear Regression (MCLLR)
approach.  Should contain algorithm description and a working example.


```python
import numpy as np
import matplotlib.pyplot as plt
from src.models.mcllr import MonteCarloLocalLinearRegression
from src.data.loader import load_fama_french
from src.data.preprocessing import time_series_split

X,y,dates = load_fama_french('data/raw/F-F_Research_Data_5_Factors_2x3.csv')
X_train,X_test,y_train,y_test,_,_ = time_series_split(X,y,dates,0.8)

model = MonteCarloLocalLinearRegression(bandwidth=1.0, n_simulations=50, subsample_size=100, random_state=0)
model.fit(X_train, y_train)

pred, std = model.predict(X_test[:5], return_std=True)
print('predictions', pred)
print('std', std)

# plot first test point uncertainty
plt.errorbar(range(len(y_test)), model.predict(X_test), yerr=model.predict(X_test, return_std=True)[1], fmt='.', alpha=0.5)
plt.title('MCLLR predictions with uncertainty')
plt.show()
```
