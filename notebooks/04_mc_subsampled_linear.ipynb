# MC Subsampled Linear

Exploration of the Monte Carlo linear regression model using subsampled
datasets.  Visualize the distribution of coefficients across simulations.

```python
import numpy as np
import matplotlib.pyplot as plt
from src.models.mc_linear import MonteCarloLinearRegression
from src.data.loader import load_fama_french
from src.data.preprocessing import time_series_split

X,y,dates = load_fama_french('data/raw/F-F_Research_Data_5_Factors_2x3.csv')
X_train,X_test,y_train,y_test,_,_ = time_series_split(X,y,dates,0.8)

mc = MonteCarloLinearRegression(n_simulations=50, subsample_size=100, random_state=1)
mc.fit(X_train, y_train)

print('Mean coefficients', mc.coefs_.mean(axis=0))
plt.boxplot(mc.coefs_)
plt.title('Coefficient variability')
plt.show()
```
