# LWLR Matrix Solution

Derivation and implementation details for solving the locally weighted linear
regression in matrix form.  Example usage goes here.

```python
import numpy as np
from src.models.lwlr import LocallyWeightedLinearRegression

# simple synthetic example
np.random.seed(0)
X = np.linspace(0,10,100).reshape(-1,1)
y = 2*X.squeeze() + np.random.normal(scale=1.0, size=X.shape[0])

lwlr = LocallyWeightedLinearRegression(bandwidth=1.0)
lwlr.fit(X,y)

# predict at a few points
x_new = np.array([[2.5],[5.0],[7.5]])
print('Predictions', lwlr.predict(x_new))
```
