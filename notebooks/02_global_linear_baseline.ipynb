# Global Linear Baseline

Outline of fitting and evaluating the global linear regression model.  Should
show coefficient estimates, residual diagnostics, etc.

```python
import sys
sys.path.append('..')

import numpy as np
from src.data.loader import load_fama_french
from src.data.preprocessing import time_series_split
from src.models.global_linear import GlobalLinearRegression

# load and split
X,y,dates = load_fama_french('../Data/raw/F-F_Research_Data_5_Factors_2x3.csv')
X_train,X_test,y_train,y_test,_,_ = time_series_split(X,y,dates,0.8)

model = GlobalLinearRegression().fit(X_train,y_train)
print('Coefficients:', model.coef_)
print('Intercept:', model.intercept_)
print('Train MSE:', model.score(X_train,y_train))
print('Test MSE:', model.score(X_test,y_test))
```

```python
# residual plot
import matplotlib.pyplot as plt

y_pred = model.predict(X_test)
plt.scatter(y_pred, y_test - y_pred, alpha=0.5)
plt.axhline(0, color='red')
plt.xlabel('Predicted')
plt.ylabel('Residual')
plt.title('Residuals vs Predicted')
plt.show()
```
